# AI Support Fabric Lab - Overview

This notebook provides an overview of the AI Support Fabric Lab architecture and demonstrates basic interactions with the lab services.

## Architecture

The lab consists of four main services:

```
[service under test] --> [telemetry_collector] --> [agentic_ai]
                                        ^             |
                                        |             v
                                      [gateway] <-- [ui_dash]
```

### Components

1. **Telemetry Collector** (Port 8081)
   - Ingests synthetic logs, metrics, and configuration events
   - Stores telemetry in SQLite database
   - Provides query API for telemetry data

2. **Agentic AI** (Port 8082)
   - Analyzes telemetry using detection rules
   - Generates findings for anomalies
   - Creates remediation plans

3. **Gateway** (Port 8080)
   - Routes requests to appropriate services
   - Adds context headers for tracing
   - Provides unified API

4. **UI Dashboard** (Port 3000)
   - Web interface for visualizing findings
   - Shows telemetry statistics
   - Displays remediation recommendations

````python
# Setup
import requests
import json
from datetime import datetime

# Gateway URL (adjust if needed)
GATEWAY_URL = "http://localhost:8080"

def pretty_print(data):
    """Pretty print JSON data"""
    print(json.dumps(data, indent=2))
````

## Check System Health

First, let's verify that all services are running properly.

````python
# Check gateway health
response = requests.get(f"{GATEWAY_URL}/health")
health_data = response.json()

print("System Health Status:")
pretty_print(health_data)
````

## Send Sample Telemetry

Let's send some sample telemetry to the collector.

````python
# Send a log entry
log_data = {
    "timestamp": datetime.utcnow().isoformat() + "Z",
    "service": "demo-service",
    "level": "INFO",
    "message": "Demo log entry from notebook",
    "request_id": "notebook-001",
    "duration_ms": 125
}

response = requests.post(
    f"{GATEWAY_URL}/api/telemetry/logs",
    json=log_data
)

print("Log Ingestion Response:")
pretty_print(response.json())
````

````python
# Send metrics
metric_data = {
    "timestamp": datetime.utcnow().isoformat() + "Z",
    "service": "demo-service",
    "metrics": {
        "cpu_percent": 35.5,
        "memory_percent": 42.3,
        "request_rate": 120,
        "error_rate": 0.5,
        "avg_latency_ms": 85
    }
}

response = requests.post(
    f"{GATEWAY_URL}/api/telemetry/metrics",
    json=metric_data
)

print("Metrics Ingestion Response:")
pretty_print(response.json())
````

## Query Telemetry

Now let's query the telemetry we've collected.

````python
# Query recent telemetry
response = requests.get(
    f"{GATEWAY_URL}/api/telemetry/query",
    params={"limit": 10}
)

telemetry_data = response.json()

print(f"Retrieved {telemetry_data['count']} telemetry entries")
print("\nFirst entry:")
if telemetry_data['telemetry']:
    pretty_print(telemetry_data['telemetry'][0])
````

## Get Telemetry Statistics

Let's check the overall telemetry statistics.

````python
# Get statistics
response = requests.get(f"{GATEWAY_URL}/api/telemetry/stats")
stats = response.json()

print("Telemetry Statistics:")
pretty_print(stats)
````
